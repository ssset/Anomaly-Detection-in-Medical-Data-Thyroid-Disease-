```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style='darkgrid', palette='dark', context='notebook')

df = pd.read_csv('cleaned_dataset_Thyroid1.csv')
df.info()
```

```python
df.isna().sum()
```

```python
binary_data = ['sex', 'on thyroxine', 'query on thyroxine', 'on antithyroid medication', 'sick', 'pregnant', 'thyroid surgery', 'I131 treatment', 
               'query hypothyroid','query hyperthyroid', 'lithium', 'goitre', 'tumor', 'hypopituitary', 'psych', 'TSH measured', 'T3 measured', 'TT4 measured', 'T4U measured',  'FTI measured']

df[(df['sex'] < 1) & (df['sex'] > 0)] = 0
n = len(binary_data)
n_cols = int(np.ceil(np.sqrt(n)))
n_rows = int(np.ceil(n / n_cols))

fig, ax = plt.subplots(ncols=n_cols, nrows=n_rows, figsize=(18, 14))
ax = ax.flatten()

for i, col in enumerate(binary_data):
    sns.countplot(data=df, x=col, hue='binaryClass', ax=ax[i])
    ax[i].set_xlabel(f'{col}')
    ax[i].set_ylabel('Binary class')
  

for i in ax[len(binary_data):]:
    i.remove()

plt.tight_layout()
plt.show()

```

```python
reg_features = ['age', 'TSH', 'TT4', 'T4U', 'FTI']

n = len(reg_features)
n_cols = int(np.ceil(np.sqrt(n)))
n_rows = int(np.ceil(n / n_cols))

fig, ax = plt.subplots(figsize=(18, 14), nrows=n_rows, ncols=n_cols)

ax = ax.flatten()
for i, col in enumerate(reg_features):
    sns.histplot(data=df, x=col, hue='binaryClass', ax=ax[i])
    plt.xlabel(f'{col}')
    plt.ylabel('Count')
    ax[i].set_yscale('log')

for i in ax[len(reg_features):]:
    i.remove()

plt.tight_layout()
plt.show()
```

```python
from itertools import combinations

reg_comb = list(combinations(reg_features, 2))

n = len(reg_comb)
n_cols = int(np.ceil(np.sqrt(n)))
n_rows = int(np.ceil(n / n_cols))

fig, ax = plt.subplots(figsize=(18, 14), ncols=n_cols, nrows=n_rows)
ax = ax.flatten()

for i, (col1, col2) in enumerate(reg_comb):
    sns.scatterplot(data=df, x=col1, y=col2, hue='binaryClass', ax=ax[i])
    ax[i].set_xlabel(f'{col1}')
    ax[i].set_ylabel(f'{col2}')

for i in ax[len(reg_comb):]:
    i.remove()

plt.tight_layout()
plt.show()
```

```python
corr = df[reg_features].corr()

plt.figure(figsize=(15, 10))
sns.heatmap(data=corr, cmap='coolwarm', annot=True, square=True)
plt.tight_layout()
plt.show()
```
